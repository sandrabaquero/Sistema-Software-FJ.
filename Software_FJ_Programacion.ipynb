```python
# IMPORTACIONES Y CONFIGURACION DEL SISTEMA

from dataclasses import dataclass
from abc import ABC, abstractmethod
import logging
from datetime import datetime

logging.basicConfig(
    filename="software_fj.log",
    level=logging.INFO,
    format="%(asctime)s - %(levelname)s - %(message)s",
    force=True
)

print("=" * 55)
print("SISTEMA SOFTWARE FJ")
print("GESTION DE CLIENTES, SERVICIOS Y RESERVAS")
print("=" * 55)

print("IMPORTACIONES REALIZADAS CORRECTAMENTE")
```

```python
# EXCEPCIONES PERSONALIZADAS

class ErrorSistema(Exception):
    pass

class ErrorCliente(ErrorSistema):
    pass

class ErrorServicio(ErrorSistema):
    pass

class ErrorReserva(ErrorSistema):
    pass


# CLASE ABSTRACTA ENTIDAD

@dataclass
class Entidad(ABC):
    identificador: str

    @abstractmethod
    def describir(self):
        pass


# CLASE CLIENTE

@dataclass
class Cliente(Entidad):
    nombre: str
    correo: str
    telefono: str

    def _post_init_(self):
        if not self.identificador.strip():
            raise ErrorCliente("Identificador obligatorio")

        if not self.nombre.strip():
            raise ErrorCliente("Nombre obligatorio")

        if "@" not in self.correo:
            raise ErrorCliente("Correo invalido")

        if not self.telefono.isdigit() or len(self.telefono) < 7:
            raise ErrorCliente("Telefono invalido")

    def describir(self):
        return "Cliente: " + self.nombre


print("EXCEPCIONES CREADAS")
print("CLASE ENTIDAD CREADA")
print("CLASE CLIENTE CREADA")
```

```python
# CLASE ABSTRACTA SERVICIO

@dataclass
class Servicio(ABC):

    codigo: str
    nombre: str
    tarifa: float

    def _post_init_(self):
        if not self.codigo or not self.nombre:
            raise ErrorServicio("Codigo y nombre obligatorios")

        if self.tarifa <= 0:
            raise ErrorServicio("La tarifa debe ser positiva")

    def calcular_total(self, subtotal, impuestos=False, descuento=0):

        if not 0 <= descuento <= 100:
            raise ErrorServicio("Descuento invalido")

        total = subtotal * (1 - descuento / 100)

        if impuestos:
            total *= 1.19

        return round(total, 2)

    @abstractmethod
    def calcular_costo(self, duracion, impuestos=False, descuento=0):
        pass

    @abstractmethod
    def describir(self):
        pass


class ReservaSala(Servicio):

    def calcular_costo(self, duracion, impuestos=False, descuento=0):
        if duracion <= 0:
            raise ErrorServicio("Duracion invalida")

        subtotal = self.tarifa * duracion
        return self.calcular_total(subtotal, impuestos, descuento)

    def describir(self):
        return "Reserva de sala: " + self.nombre


class AlquilerEquipo(Servicio):

    def calcular_costo(self, duracion, impuestos=False, descuento=0):
        if duracion <= 0:
            raise ErrorServicio("Duracion invalida")

        subtotal = self.tarifa * duracion
        return self.calcular_total(subtotal, impuestos, descuento)

    def describir(self):
        return "Alquiler de equipo: " + self.nombre


class Consultoria(Servicio):

    def calcular_costo(self, duracion, impuestos=False, descuento=0):
        if duracion <= 0:
            raise ErrorServicio("Duracion invalida")

        subtotal = self.tarifa * duracion
        return self.calcular_total(subtotal, impuestos, descuento)

    def describir(self):
        return "Consultoria especializada: " + self.nombre


print("CLASE ABSTRACTA SERVICIO CREADA")
print("SERVICIO DE SALAS CREADO")
print("SERVICIO DE EQUIPOS CREADO")
print("SERVICIO DE CONSULTORIA CREADO")
```

```python
# LISTAS INTERNAS DEL SISTEMA

clientes = []
servicios = []
reservas = []

def registrar_cliente(cliente):
    clientes.append(cliente)
    logging.info("Cliente registrado: %s", cliente.nombre)

def registrar_servicio(servicio):
    servicios.append(servicio)
    logging.info("Servicio registrado: %s", servicio.nombre)

def registrar_reserva(reserva):
    reservas.append(reserva)
    logging.info("Reserva registrada: %s", reserva.codigo)

print("LISTAS DEL SISTEMA CREADAS")
```

```python
# CLASE RESERVA

@dataclass
class Reserva:
    codigo: str
    cliente: Cliente
    servicio: Servicio
    duracion: int
    estado: str = "Pendiente"

    def _post_init_(self):
        if not self.codigo.strip():
            raise ErrorReserva("Codigo obligatorio")

        if self.duracion <= 0:
            raise ErrorReserva("La duracion debe ser positiva")

    def confirmar(self):
        self.estado = "Confirmada"
        logging.info("Reserva %s confirmada", self.codigo)

    def cancelar(self):
        self.estado = "Cancelada"
        logging.info("Reserva %s cancelada", self.codigo)

    def procesar(self):
        total = self.servicio.calcular_costo(self.duracion)
        return total

    def describir(self):
        return (
            "Reserva " + self.codigo
            + " - Cliente: " + self.cliente.nombre
            + " - Servicio: " + self.servicio.nombre
            + " - Estado: " + self.estado
        )

print("CLASE RESERVA CREADA CORRECTAMENTE")
```

```python
# REGISTRO DE CLIENTES

cliente1 = Cliente("C001", "Laura Gomez", "laura@gmail.com", "3001234567")
cliente2 = Cliente("C002", "Carlos Perez", "carlos@gmail.com", "3101234567")

registrar_cliente(cliente1)
registrar_cliente(cliente2)

# REGISTRO DE SERVICIOS

servicio1 = ReservaSala("S001", "Sala de reuniones", 50000)
servicio2 = AlquilerEquipo("S002", "Video beam", 30000)
servicio3 = Consultoria("S003", "Consultoria tecnologica", 80000)

registrar_servicio(servicio1)
registrar_servicio(servicio2)
registrar_servicio(servicio3)

print("CLIENTES Y SERVICIOS REGISTRADOS")
print("Total clientes:", len(clientes))
print("Total servicios:", len(servicios))
```

```python
# CREACION DE RESERVAS

reserva1 = Reserva("R001", cliente1, servicio1, 2)
reserva2 = Reserva("R002", cliente2, servicio2, 3)
reserva3 = Reserva("R003", cliente1, servicio3, 1)

registrar_reserva(reserva1)
registrar_reserva(reserva2)
registrar_reserva(reserva3)

print("RESERVAS REGISTRADAS")
print("Total de reservas:", len(reservas))

for reserva in reservas:
    print(reserva.describir())
```

```python
# PRUEBA DE CALCULO DE COSTOS

print("CALCULO DE COSTOS DE LOS SERVICIOS")

for reserva in reservas:
    total = reserva.procesar()
    print(reserva.codigo, "-", reserva.servicio.nombre, "- Total: $", total)

print("--------------------------------")
print("PRUEBA DE COSTOS FINALIZADA")
```

```python
# PRUEBA DE CONFIRMACION Y CANCELACION

reserva1.confirmar()
reserva2.confirmar()
reserva3.cancelar()

print("ESTADOS DE LAS RESERVAS")

for reserva in reservas:
    print(reserva.codigo, "-", reserva.estado)

print("PRUEBA DE ESTADOS FINALIZADA")
```

```python
# PRUEBA DE IMPUESTOS Y DESCUENTOS

print("CALCULO CON IMPUESTOS Y DESCUENTOS")

print("Sala con IVA:",
      servicio1.calcular_costo(2, impuestos=True))

print("Video beam con descuento:",
      servicio2.calcular_costo(3, descuento=10))

print("Consultoria con IVA y descuento:",
      servicio3.calcular_costo(1, impuestos=True, descuento=5))

print("PRUEBA DE IMPUESTOS Y DESCUENTOS FINALIZADA")
```

```python
# PRUEBAS DE VALIDACION Y MANEJO DE ERRORES

print("INICIO DE PRUEBAS DE ERRORES")

# Prueba 1: cliente con correo incorrecto
try:
    cliente_malo = Cliente(
        "C003", "Pedro Ruiz", "correo_invalido", "3001234567"
    )
    registrar_cliente(cliente_malo)
except ErrorCliente as error:
    print("Error de cliente detectado:", error)
    logging.error("Error de cliente: %s", error)

# Prueba 2: servicio con tarifa negativa
try:
    servicio_malo = ReservaSala(
        "S004", "Sala incorrecta", -5000
    )
except ErrorServicio as error:
    print("Error de servicio detectado:", error)
    logging.error("Error de servicio: %s", error)

# Prueba 3: reserva con duración cero
try:
    reserva_mala = Reserva(
        "R004", cliente1, servicio1, 0
    )
except ErrorReserva as error:
    print("Error de reserva detectado:", error)
    logging.error("Error de reserva: %s", error)

print("EL SISTEMA CONTINUA FUNCIONANDO")
print("PRUEBAS DE ERRORES FINALIZADAS")
```

```python
# BUSQUEDA DE CLIENTES

def buscar_cliente(identificador):
    for cliente in clientes:
        if cliente.identificador == identificador:
            return cliente
    raise ErrorCliente("Cliente no encontrado")

try:
    encontrado = buscar_cliente("C001")
    print("CLIENTE ENCONTRADO:")
    print(encontrado.describir())

except ErrorCliente as error:
    print("ERROR:", error)

try:
    buscar_cliente("C999")
except ErrorCliente as error:
    print("BUSQUEDA INCORRECTA DETECTADA:", error)

print("PRUEBA DE BUSQUEDA FINALIZADA")
```

```python
# MANEJO AVANZADO DE EXCEPCIONES

def procesar_reserva_segura(reserva):
    try:
        total = reserva.procesar()

    except ErrorSistema as error:
        logging.exception("Error al procesar reserva")
        raise ErrorReserva(
            "No fue posible procesar la reserva"
        ) from error

    else:
        print("Reserva procesada correctamente")
        print("Valor total: $", total)

    finally:
        print("Proceso de reserva finalizado")


# PRUEBA EXITOSA
print("PRIMERA PRUEBA")
procesar_reserva_segura(reserva1)

# PRUEBA CON ERROR
print("SEGUNDA PRUEBA")

duracion_original = reserva2.duracion
reserva2.duracion = 0

try:
    procesar_reserva_segura(reserva2)
except ErrorReserva as error:
    print("EXCEPCION CONTROLADA:", error)

reserva2.duracion = duracion_original

print("EL SISTEMA SIGUE FUNCIONANDO")
```
